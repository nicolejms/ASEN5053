```python
from astropy import units as u

from poliastro.bodies import Earth, Jupiter
from poliastro.plotting.porkchop import PorkchopPlotter
from poliastro.util import time_range

launch_span = time_range("2024-07-01", end="2024-10-01")
arrival_span = time_range("2026-06-01", end="2028-01-01")

porkchop_plot = PorkchopPlotter(Earth, Jupiter, launch_span, arrival_span,ax=None, tfl=True, vhp=True, max_c3=100.0 * u.km ** 2 / u.s ** 2, max_vhp=5 * u.km / u.s)
dv_dpt, dv_arr, c3dpt, c3arr, tof = porkchop_plot.porkchop()
```
